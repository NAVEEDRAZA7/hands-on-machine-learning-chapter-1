## — Load Dataset

### Extract the Input Feature (GDP per Capita)
### Extract the Target Variable (Life Satisfaction)

```python
data_root = "https://github.com/ageron/data/raw/main/"                             # # Base URL where the dataset is stored

lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")                           # # Download and load the CSV file into a Pandas DataFrame

X = lifesat[["GDP per capita (USD)"]].values                                       # # Select GDP as the input feature (X)

y = lifesat[["Life satisfaction"]].values                                          # # Select Life Satisfaction as the target (y)
```

```python
model = LinearRegression()      # # Create a Linear Regression model
```

```python
model.fit(X, y)          # # Learn relationship between GDP and Life Satisfaction
```

### — Creating Folder & Downloading Files (Datasets from internet) using code

```python
from pathlib import Path               # # Work with folders/files easily

import urllib.request                  # # Download files from the Internet

datapath = Path() / "datasets" / "lifesat"
# # Create path: datasets/lifesat

datapath.mkdir(parents=True, exist_ok=True)
# # Create folder if it doesn't already exist

data_root = "https://github.com/ageron/data/raw/main/"
# # Base URL for all datasets

for filename in ("oecd_bli.csv", "gdp_per_capita.csv"):
    # # Loop through required dataset names

    if not (datapath / filename).is_file():
        # # Download only if the file doesn't exist

        print("Downloading", filename)

        url = data_root + "lifesat/" + filename
        # # Build complete download URL

        urllib.request.urlretrieve(url, datapath / filename)
        # # Download and save the file
```

### — Read CSV Files

```python
oecd_bli = pd.read_csv(datapath / "oecd_bli.csv")                       # # Load OECD Better Life Index dataset

gdp_per_capita = pd.read_csv(datapath / "gdp_per_capita.csv")           # # Load GDP dataset
```

### — Data Cleaning : Filter the GDP Dataset to Keep Only the Year 2020

```python
gdp_year = 2020                         # # Select year

gdppc_col = "GDP per capita (USD)"      # # New GDP column name

lifesat_col = "Life satisfaction"       # # Target column name

gdp_per_capita = gdp_per_capita[
    gdp_per_capita["Year"] == gdp_year
]
# # Keep only 2020 records

gdp_per_capita = gdp_per_capita.drop(
    ["Code", "Year"], axis=1
)
# # Remove unnecessary columns

gdp_per_capita.columns = [
    "Country",
    gdppc_col
]
# # Rename columns

gdp_per_capita.set_index("Country", inplace=True)
# # Make Country the row index
```

### — Prepare OECD Dataset to Keep Only the Total Inequality Records

```python
oecd_bli = oecd_bli[
    oecd_bli["INEQUALITY"] == "TOT"
]
# # Keep only total inequality records

oecd_bli = oecd_bli.pivot(
    index="Country",
    columns="Indicator",
    values="Value"
)
# # Convert long table into wide format
```

### — Merge Datasets Merge GDP and Life Satisfaction Datasets

```python
# # Merge datasets using Country index

full_country_stats = pd.merge(
    left=oecd_bli,
    right=gdp_per_capita,
    left_index=True,
    right_index=True
)

# # Sort countries by GDP

full_country_stats.sort_values(
    by=gdppc_col,
    inplace=True
)

# # Keep only the two required columns

full_country_stats = full_country_stats[
    [gdppc_col, lifesat_col]
]
```

### — Create Training Dataset : Select Countries Within the Desired GDP Range

```python
min_gdp = 23_500                  # # Minimum GDP to include

max_gdp = 62_500                  # # Maximum GDP to include

# # Keep only countries within the selected GDP range

country_stats = full_country_stats[
    (full_country_stats[gdppc_col] >= min_gdp) &
    (full_country_stats[gdppc_col] <= max_gdp)
]

country_stats.head()              # # Display the first 5 rows
```

### — Save Clean Dataset

```python
country_stats.to_csv(datapath / "lifesat.csv")
# # Save cleaned training dataset

full_country_stats.to_csv(datapath / "lifesat_full.csv")
# # Save complete dataset
```

### — Train the Linear Regression Model to Find the Best-Fit Line

```python
# # Import Linear Regression module
from sklearn import linear_model

# # Select GDP as input feature
X_sample = country_stats[[gdppc_col]].values

# # Select Life Satisfaction as target
y_sample = country_stats[[lifesat_col]].values

# # Create a Linear Regression model
lin1 = linear_model.LinearRegression()

# # Train the model
lin1.fit(X_sample, y_sample)

# # Extract intercept (θ₀)
t0 = lin1.intercept_[0]

# # Extract slope (θ₁)
t1 = lin1.coef_.ravel()[0]

# # Print learned parameters
print(f"θ₀ = {t0:.2f}")
print(f"θ₁ = {t1:.2e}")
```

### — Draw the Best-Fit Regression Line

```python
# # Plot the training dataset
country_stats.plot(
    kind="scatter",
    figsize=(5, 3),
    grid=True,
    x=gdppc_col,
    y=lifesat_col
)

# # Generate X values
X = np.linspace(min_gdp, max_gdp, 1000)

# # Draw the learned regression line
plt.plot(X, t0 + t1 * X, "b")

# # Display intercept value
plt.text(
    max_gdp - 20000,
    min_life_sat + 1.9,
    fr"$\theta_0={t0:.2f}$",
    color="b"
)

# # Display slope value
plt.text(
    max_gdp - 20000,
    min_life_sat + 1.3,
    fr"$\theta_1={t1*1e5:.2f}\times10^{{-5}}$",
    color="b"
)

# # Set graph limits
plt.axis([min_gdp, max_gdp, min_life_sat, max_life_sat])

# # Display graph
plt.show()
```
